```python
def _safe_min_last_k(obj, k=20):
    """
    Return the minimum of the *last k* numeric entries inside `obj`.
    Works whether `obj` is
      • a list of floats,
      • a stringified list (CSV sometimes stores lists as strings),
      • nested lists such as [None, [v11,v12,…], …],
      • or None / NaN.
    If no numeric data are found, returns NaN.
    """
    # Convert "string that looks like a list" → Python list
    if isinstance(obj, str) and obj.startswith("["):
        try:
            obj = ast.literal_eval(obj)
        except Exception:
            return np.nan

    # Flatten any nested structure
    flat = []
    def _flatten(x):
        if x is None or (isinstance(x, float) and np.isnan(x)):
            return
        if isinstance(x, (list, tuple, np.ndarray)):
            for xi in x: _flatten(xi)
        else:
            flat.append(x)
    _flatten(obj)

    if not flat:
        return np.nan
    flat = np.asarray(flat, dtype=float)
    return np.min(flat[-k:])


df = pd.read_csv("./qaoa_stats.csv")

# Keep only rows that *actually have* post-LS data
df = df[df["step4_iter_best_fx"].notna()].copy()

k = 20
best_col = f"step4_best_fx_{k}iters"
gap_col  = f"step4_rel_gap_{k}iters"

# Build the helper columns if they are missing or stale
if best_col not in df.columns:
    df[best_col] = df["step4_iter_best_fx"].apply(_safe_min_last_k, k=k)

if gap_col not in df.columns:
    df[gap_col] = (df[best_col] - df["refvalue"]) / df["refvalue"]


step3_gap_median = df["step3_rel_gap"].median(skipna=True)
step4_gap_median = df[gap_col].median(skipna=True)

print("Columns:")
print(df.columns.tolist())

print(f"\nMedian step-3 gap  : {step3_gap_median:.4%}")
print(f"Median step-4 gap  : {step4_gap_median:.4%}")

# ---------- meta columns (depth / timing) ------------------------------------
depth_cols = [c for c in df.columns if "depth" in c.lower()]
time_cols  = [c for c in df.columns if "time"  in c.lower()]

print("\nDepth-like columns :", depth_cols or "— none found —")
print("Time-like columns  :", time_cols  or "— none found —")
```
